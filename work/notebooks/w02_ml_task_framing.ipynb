## My Lane as an ML Task (Type)

**Chosen Lane:** Refresh / Content Opportunity Scoring

**ML Task Type:** Scoring (with Ranking)

I chose **Refresh / Content Opportunity Scoring**, which is primarily a **scoring** task. The goal is to assign each webpage a **Refresh Opportunity Score** based on its search performance, engagement, and content characteristics. The pages are then **ranked** from highest to lowest score so content teams can focus on the pages with the greatest potential for improvement. This is not a classification task because pages are not simply labeled as "refresh" or "do not refresh"; instead, each page receives a score that supports prioritization.

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("/content/sample_data/content_refresh_anonymized.csv")

print("ML Task Type : Scoring (Ranking)")
print("Unit of Analysis : One webpage")
print("Dataset Shape :", df.shape)

# Display a few rows
df.head()
```

## Target or Proxy

**Target:** Refresh Opportunity Score

This project predicts a **Refresh Opportunity Score** for each webpage. Since there is no direct label that says whether a page should be refreshed, the target is a **proxy** rather than an observed outcome.

The proxy score is derived from existing search performance, engagement, and content-related signals such as impressions, CTR, content age, and other available features. Pages with lower performance and higher refresh potential are expected to receive higher scores, helping content teams prioritize their review efforts.

```python


df["refresh_opportunity_score"] = (
    df["content_age_days"] * (1 - df["ctr"])
)

df[["content_age_days", "ctr", "refresh_opportunity_score"]].head()
```

## Success Metric

**Success Metric:** Precision@10

I chose **Precision@10** as the primary success metric because the goal is to rank webpages that should be reviewed first. Precision@10 measures how many of the top 10 recommended pages are actually valuable refresh opportunities. A higher Precision@10 means the model is helping content teams spend their time on the most important pages.

If ground-truth labels are not available, the quality of the recommendations can also be evaluated through expert review by SEO or content teams.

```python
# Example: Number of pages in the dataset
print("Total webpages:", len(df))

print("\nExample Success Metric")
print("Metric: Precision@10")
print("Goal: Maximize the number of truly high-opportunity pages in the Top 10 recommendations.")
```

## Why ML Beats a Fixed Rule Here

A fixed rule such as "refresh every page older than 365 days" or "refresh pages with CTR below 1%" is too simple because webpage performance depends on multiple factors at the same time. A page may be old but still perform well, while a newer page may have low engagement and high improvement potential.

Machine learning can consider several features together, such as impressions, CTR, content age, engagement, and other search signals, to estimate a Refresh Opportunity Score. This allows the model to identify more meaningful patterns than a single if-statement or threshold, making the recommendations more useful for content teams.

```python
# Example features that could be used together
features = [
    "impressions_90d",
    "ctr",
    "content_age_days"
]

available_features = [f for f in features if f in df.columns]

print("Example features used together by an ML model:")
print(available_features)
```
